```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import matplotlib.patches as patches
import matplotlib as mpl

# Font configuration
mpl.rcParams.update({
    "font.family": "serif",
    "mathtext.fontset": "cm"
})

# Load PNG images
img_start_s1 = mpimg.imread('scenarios/H1I.png')
img_start_s2 = mpimg.imread('scenarios/H2I.png')
img_end_s1   = mpimg.imread('scenarios/H1F.png')
img_end_s2   = mpimg.imread('scenarios/H2F.png')

# Create figure and subplots
fig, axes = plt.subplots(2, 2, figsize=(8, 6))
fig.subplots_adjust(wspace=0.02) 

# Show images
axes[0, 0].imshow(img_start_s1)
axes[0, 1].imshow(img_end_s1)
axes[1, 0].imshow(img_start_s2)
axes[1, 1].imshow(img_end_s2)

# Turn off axes
for ax in axes.flatten():
    ax.axis('off')

# Add row labels
fig.text(0.18, 0.7, 'Scenario 1', va='center', ha='right', rotation='vertical', fontsize=16, fontweight='bold')
fig.text(0.18, 0.27, 'Scenario 2', va='center', ha='right', rotation='vertical', fontsize=16, fontweight='bold')

start_ax = axes[0, 0]
end_ax   = axes[0, 1]

# Get center x-coordinates of each column in figure coordinates
start_center = start_ax.get_position().x0 + start_ax.get_position().width / 2
end_center   = end_ax.get_position().x0 + end_ax.get_position().width / 2

# Add column labels centered over each column
fig.text(start_center+0.05, 0.9, 'Start', ha='center', fontsize=16, fontweight='bold')
fig.text(end_center+0.03,   0.9, 'End', ha='center', fontsize=16, fontweight='bold')

fig.text(end_center+0.138,   0.825, 'TT', ha='center', fontsize=16)
fig.text(end_center+0.138,   0.41, 'HB', ha='center', fontsize=16)


# Add vertical "HOMING" title on the left
fig.text(0.1, 0.5, 'HOMING', va='center', ha='center', rotation='vertical', fontsize=14, fontweight='bold')

# Set background colors
fig.patch.set_facecolor('lightgray')
for ax in axes.flatten():
    ax.set_facecolor('whitesmoke')

# Add arrows between Start and End columns (figure coordinates)
# Arrow for top row
arrow1 = patches.FancyArrow(
    0.515, 0.66,  # (x, y) start
    0.07, 0.0,   # (dx, dy)
    transform=fig.transFigure,
    width=0.01,
    head_width=0.03,
    head_length=0.02,
    length_includes_head=True,
    color='black'
)
fig.patches.append(arrow1)

# Arrow for bottom row
arrow2 = patches.FancyArrow(
    0.515, 0.26,  # (x, y) start
    0.07, 0.0,   # (dx, dy)
    transform=fig.transFigure,
    width=0.01,
    head_width=0.03,
    head_length=0.02,
    length_includes_head=True,
    color='black'
)
fig.patches.append(arrow2)

# Layout adjustment
plt.tight_layout(rect=[0.1, 0.05, 1, 0.9])

# Save to PDF
fig.savefig("ALIFE_images/homing_layout.pdf", format='pdf', bbox_inches='tight')

plt.show()

```

```python
from pdf2image import convert_from_path
import matplotlib.pyplot as plt
import matplotlib as mpl

# Font configuration
mpl.rcParams.update({
    "font.family": "serif",
    "mathtext.fontset": "cm"
})

# Convert PDF to image (returns a list of images, one per page)
images = convert_from_path("M-A1-A2.pdf")
img = images[0]  # Assuming single-page PDF

# Create figure
fig, ax = plt.subplots(figsize=(10, 4))
ax.imshow(img)
ax.axis('off')

# Add labels
fig.text(0.08, 0.6, 'AGGREGATION', va='center', ha='center', rotation='vertical', fontsize=14, fontweight='bold')
fig.text(0.34, 0.95, 'Scenario 1', ha='center', fontsize=16)
fig.text(0.7, 0.95, 'Scenario 2', ha='center', fontsize=16)

# Save final layout

plt.tight_layout()

fig.savefig("alife_results/agg_performance.pdf", format='pdf', bbox_inches='tight')
plt.show()
```
