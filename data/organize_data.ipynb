```python
import json
```

```python
def read_txt_file(file_path):
    """Lee un archivo de texto y devuelve una lista con cada línea como un elemento."""
    with open(file_path, 'r', encoding='utf-8') as file:
        lines = [line.strip() for line in file.readlines() if line.strip()]
    return lines

def load_existing_json(output_file):
    """Carga el JSON existente si existe, de lo contrario, retorna una estructura vacía."""
    try:
        with open(output_file, 'r', encoding='utf-8') as json_file:
            return json.load(json_file)
    except (FileNotFoundError, json.JSONDecodeError):
        return {"missions": []}

def organize_data(file_path, mission_name, output_file):
    """Organiza los datos y añade una nueva misión conservando las anteriores."""
    data = read_txt_file(file_path)
    existing_data = load_existing_json(output_file)
    
    new_mission = {mission_name: data}
    existing_data["missions"].append(new_mission)
    
    return existing_data

def save_to_json(data, output_file):
    """Guarda los datos en un archivo JSON."""
    with open(output_file, 'w', encoding='utf-8') as json_file:
        json.dump(data, json_file, indent=4, ensure_ascii=False)
```

```python
input_txt = "/home/robotmaster/Downloads/controllers-folder/hm-cm-100/fsm/fsm.txt"  # Nombre del archivo de entrada
output_json = "missions.json"  # Nombre del archivo de salida
mission_name = "hm-100"  # Nombre de la misión

organized_data = organize_data(input_txt, mission_name, output_json)
save_to_json(organized_data, output_json)
print(f"Archivo JSON generado: {output_json}")
```
